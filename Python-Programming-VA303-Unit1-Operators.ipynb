## **Arithmetic Operators**

Arithmetic operators are used to perform mathematical calculations.

<b>Operator	&nbsp;&nbsp;  Description &nbsp;&nbsp;  Example </b> <br>
+	&nbsp;&nbsp;&nbsp;Addition	       &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;10 + 5 <br>
-	&nbsp;&nbsp;&nbsp;Subtraction	       &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;10 - 5 <br>
*	&nbsp;&nbsp;&nbsp;Multiplication	   &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;10 * 5 <br>
/	&nbsp;&nbsp;&nbsp;Division	       &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;10 / 5 <br> 
//	&nbsp;&nbsp;&nbsp;Floor Division	   &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;10 // 3 <br>
%	&nbsp;&nbsp;&nbsp;Modulus	           &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;10 % 3 <br>
**	&nbsp;&nbsp;&nbsp;Exponent	       &nbsp;&nbsp;&nbsp;&nbsp;&nbsp;2 ** 5 <br>

```python
# Example: Basic Arithmetic
a = 25
b = 6

print("Addition =", a + b)
print("Subtraction =", a - b)
print("Multiplication =", a * b)
print("Division =", a / b)
print("Floor Division =", a // b)
print("Modulus =", a % b)
print("Power =", a ** b)
```

## **Relational (Comparison) Operators**

Relational operators compare two values and return True or False.

<b>Operator &nbsp;&nbsp;	Description  </b> <br>
==	&nbsp;&nbsp;&nbsp;&nbsp;        Equal to      <br>
!=	&nbsp;&nbsp;&nbsp;&nbsp;        Not Equal     <br>
'>'	 &nbsp;&nbsp;&nbsp;&nbsp;       Greater Than  <br>
'<'	  &nbsp;&nbsp;&nbsp;&nbsp;      Less Than     <br>
'>='  &nbsp;&nbsp;&nbsp;&nbsp;    Greater Than or Equal <br>
'<='  &nbsp;&nbsp;&nbsp;&nbsp;	    Less Than or Equal  <br>

```python
# Example
a = 20
b = 15

print(a == b)
print(a != b)
print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
```

## **Logical Operators**

Logical operators combine multiple conditions.

Operator &nbsp;&nbsp;	Description  <br>
and	&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; Returns True if both conditions are True  <br>
or	&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; Returns True if at least one condition is True  <br>
not	&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp; Reverses the Boolean value <br>

```python
# Example
age = 21
cgpa = 8.5

print(age >= 18 and cgpa >= 8)
print(age >= 18 or cgpa >= 9)
print(not(age >= 18))
```

## **Bitwise Operators**

Bitwise operators work on the binary representation of integers.

<b>Operator &nbsp;&nbsp;	Description </b><br>
&	&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Bitwise AND  <br>
|	&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Bitwise OR   <br>
^	&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Bitwise XOR  <br>
~	&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Bitwise NOT  <br>
'<<'	&nbsp;&nbsp;&nbsp;&nbsp;Left Shift  <br>
'>>'	&nbsp;&nbsp;&nbsp;&nbsp;Right Shift  <br>

```python
## Example
a = 10
b = 6
x = 5

# Bitwise AND
print(a & b)

# Bitwise OR
print(a | b)

# Bitwise XOR
print(a ^ b)

# Left Shift
print(x << 1)
print(x << 2)

# Right Shift
print(x >> 1)
print(x >> 2)

# Bitwise NOT
print(~x)
```

## **Membership Operators**

Membership operators check whether a value exists in a sequence.

<b>Operator&nbsp;&nbsp;	Description</b>  <br>
in	&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Returns True if value exists <br>
not in	&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;Returns True if value does not exist <br>

```python
# Example with String
language = "Python Programming"

print("Python" in language)
print("Java" in language)
```

```python
# Example with List
numbers = [10, 20, 30, 40, 50]

print(30 in numbers)
print(100 in numbers)
```

## **Problem Statement**

A university offers a merit scholarship based on the following criteria:

CGPA ≥ 8.0  <br>
Attendance ≥ 75% <br>
Family income ≤ ₹500,000 <br>
No disciplinary action <br>

The program should: <br>

Accept user inputs. <br>
Use arithmetic, relational, and logical operators. <br>
Display scholarship eligibility. <br>

## **Solution**

```python
print("===== Scholarship Eligibility Checker =====")

cgpa = float(input("Enter CGPA: "))
attendance = float(input("Enter Attendance (%): "))
family_income = float(input("Enter Annual Family Income (₹): "))
disciplinary_action = input("Any disciplinary action? (yes/no): ").strip().lower()

eligible = (
    cgpa >= 8.0 and
    attendance >= 75 and
    family_income <= 500000 and
    disciplinary_action == "no"
)

print("\n===== Result =====")
print("CGPA:", cgpa)
print("Attendance:", attendance)
print("Family Income:", family_income)
print("Disciplinary Action:", disciplinary_action)
print("Scholarship Eligible:", eligible)
```
