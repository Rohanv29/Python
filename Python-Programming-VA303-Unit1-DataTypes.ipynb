## **Integer (int)**

Integers are whole numbers without a decimal point.

Examples:

```python
a = 25
b = -150
c = 0

print(a)
print(b)
print(c)
```

## **Arithmetic Operations**

```python
x = 15
y = 4

print("Addition =", x + y)
print("Subtraction =", x - y)
print("Multiplication =", x * y)
print("Division =", x / y)
print("Floor Division =", x // y)
print("Modulus =", x % y)
print("Power =", x ** y)
```

## **Complex Numbers (complex)**

Complex numbers contain: <br>
Real Part <br>
Imaginary Part

Format:
a + bj

**Example**

```python
z = 4 + 3j

print(z)
print(type(z))
```

## **Real and Imaginary Parts**

```python
z = 8 + 5j

print(z.real)
print(z.imag)
```

## **Complex Arithmetic**

```python
a = 3 + 2j
b = 4 + 5j

print(a + b)
print(a - b)
print(a * b)  # (Formula to calculate (a+bj)*(c+dj) is (ac-bd) + (ad + bc)j
print(a / b)  # (Formula to calculate (a+bj)/(c+dj) is ((ac+bd) + (-ad+bc)j) / (c**2+d**2)
```

## **Creating Complex Numbers**

```python
x = complex(10, 5)

print(x)
```

## **Boolean (bool)**

Boolean has only two values.

True
False

## **Creating Boolean Variables**

```python
is_student = True
is_passed = False

print(is_student)
print(is_passed)
```

## **Comparison Operators**

```python
a = 10
b = 20

print(a > b)
print(a < b)
print(a == b)
print(a != b)
```

## **Logical Operators**

```python
age = 20
cgpa = 8.5

print(age > 18 and cgpa > 8)
print(age > 18 or cgpa > 9)
print(not(age > 18))
```

## **Boolean from Numbers**

```python
print(bool(0))
print(bool(1))
print(bool(-10))
```

## **Boolean from Strings**

```python
print(bool(""))
print(bool("Python"))
```
