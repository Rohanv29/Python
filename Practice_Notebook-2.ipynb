## **Challenge Version of Question 1 (Maximum 7 Attempts)**

## **Your Solution**

```python
# Write your code here
import random

number = random.randint(1, 100)
attempts = 0

while attempts < 7:
    guess = int(input("Enter your guess (1-100): "))
    attempts += 1

    if guess == number:
        print("Congratulations! You guessed the correct number.")
        print("You guessed it in", attempts, "attempt(s).")
        break
    elif guess > number:
        print("Your guess is too high.")
    else:
        print("Your guess is too low.")

if attempts == 7 and guess != number:
    print("Game Over! You have used all 7 attempts.")
    print("The correct number was:", number)
```

## **Question 4:**

Bank PIN Verification – Allow only 3 attempts using a while loop and break.

## **Your Solution**

```python
# Write your code here
correct_pin = "1234"
attempts = 0

while attempts < 3:
    pin = input("Enter PIN: ")

    if pin == correct_pin:
        print("PIN Verified")
        break
    else:
        attempts += 1
        print("Incorrect PIN")

if attempts == 3:
    print("Card Blocked")
```

## **Question 5:**

Write a system that forces a user to enter the correct password "Python123".  <br>
The user has a maximum of 3 attempts.  <br>
If they succeed, print "Access Granted".  <br>
If they run out of attempts, print "Account Locked". <br>

## **Your Solution**

```python
# Write your code here
password = "Python123"
attempts = 0

while attempts < 3:
    user = input("Enter Password: ")

    if user == password:
        print("Access Granted")
        break
    else:
        attempts += 1
        print("Wrong Password")

if attempts == 3:
    print("Account Locked")
```

## **Question 6:**

Mini Banking Menu – Create a menu-driven application (Deposit, Withdraw, Balance Inquiry, Exit) using a while loop and conditional statements.

## **Your Solution**

```python
# Write your code here
balance = 10000

while True:
    print("\n1. Deposit")
    print("2. Withdraw")
    print("3. Balance Inquiry")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        amount = float(input("Enter deposit amount: "))
        balance += amount
        print("Amount Deposited Successfully")

    elif choice == 2:
        amount = float(input("Enter withdrawal amount: "))
        if amount <= balance:
            balance -= amount
            print("Withdrawal Successful")
        else:
            print("Insufficient Balance")

    elif choice == 3:
        print("Current Balance:", balance)

    elif choice == 4:
        print("Thank You!")
        break

    else:
        print("Invalid Choice")
```

## **Question 7**

Write a program that prints numbers from 1 to 20. <br>
For multiples of 3, print "Fizz" instead of the number.  <br>
For multiples of 5, print "Buzz".          <br>
For numbers which are multiples of both 3 and 5, print "FizzBuzz". <br>

## **Your Solution**

```python
# Write your code here
for i in range(1, 21):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

## **Question 8**

Write an interactive calculator that continuously asks the user for numbers until they type 0. 
Once 0 is typed, display the total count of numbers entered, their average, and the largest number provided (excluding the 0).

## **Your Solution**

```python
# Write your code here
count = 0
total = 0
largest = None

while True:
    num = int(input("Enter a number (0 to stop): "))

    if num == 0:
        break

    count += 1
    total += num

    if largest is None or num > largest:
        largest = num

if count > 0:
    average = total / count
    print("Count:", count)
    print("Average:", average)
    print("Largest Number:", largest)
else:
    print("No numbers entered.")
```
